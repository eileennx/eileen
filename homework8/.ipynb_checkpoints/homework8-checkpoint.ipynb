```python
# 3.
df = pd.read_csv('../../course_assignments/homework8/state_data.csv')
```

```python
# 4. 
shape = df.shape
print(f"The shape of the data is: {shape}")
print(f"There are {shape[1]} columns and {shape[0]} rows.")
```

```python
# 5.
print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

'''
- The first column represents the state and its abbreviation, the second column represents the coordinate of the north-south position of the state's center, the third column represents the coordinate of the east-west position of the state's center, and the last column represents the full name of the state.
- The first state is Alaska
- The last state is Wyoming
- There are 52 entries.
'''
```

```python
# 6. 
med_lat = df['latitude'].median()
med_long = df['longitude'].median()
print(f"Median Latitude: {med_lat:.2f}")
print(f"Median Longitude: {med_long:.2f}")

# 7. 
df['lat_dist'] = (df['latitude'] - med_lat).abs()

# 8.
df['long_dist'] = (df['longitude'] - med_long).abs()
```

```python
# 9.
new_shape = df.shape
print(f"New shape: {new_shape}")
```

```python
# 10. 
closest_lat_idx = df['lat_dist'].idxmin()
farthest_lat_idx = df['lat_dist'].idxmax()

closest_long_idx = df['long_dist'].idxmin()
farthest_long_idx = df['long_dist'].idxmax()

print(f"State closest to median latitude: {df.loc[closest_lat_idx, 'state']} by {df.loc[closest_lat_idx, 'lat_dist']:.2f}")
print(f"State farthest from median latitude: {df.loc[farthest_lat_idx, 'state']} by {df.loc[farthest_lat_idx, 'lat_dist']:.2f}")
print(f"State closest to median longitude: {df.loc[closest_long_idx, 'state']} by {df.loc[closest_long_idx, 'long_dist']:.2f}")
print(f"State farthest from median longitude: {df.loc[farthest_long_idx, 'state']} by {df.loc[farthest_long_idx, 'long_dist']:.2f}")
```

```python
# 11.
top_3_north = df.nlargest(3, 'latitude')
print(top_3_north)
'''
- I found this under the "Computations / descriptive stats" section on the website.
- This function returns the n (here, 3) rows with the largest values in a specific column (here, 'latitude'). It is useful because it's much faster than sorting the whole dataframe and then using .head().
'''
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 4.
df = sns.load_dataset('exercise')

# 5.
df_cleaned = df.dropna(axis=1)

# 6.
print(f"Shape of the dataset: {df.shape}")
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")

print("\nFirst five rows:")
display(df.head())

print("\nLast five rows:")
display(df.tail())

'''
Columns:
- id: Unique ID for each participant
- diet: Categorical (low fat vs no fat)
- pulse: Heart rate in beats per minute (bpm)
- time: Duration of exercise (1, 15, or 30 min).
- kind: Intensity of exercise (rest, walking, running)
'''
```

```python
# 7 & 8
plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)
sns.scatterplot(data=df, x='id', y='pulse', hue='kind', palette='viridis')
plt.title('Heart Rate (bpm) by Participant ID')
plt.xlabel('Participant ID')
plt.ylabel('Pulse (bpm)')
plt.legend(title='Activity Type')

plt.subplot(2, 2, 2)
sns.scatterplot(data=df, x='time', y='pulse', color='darkorange', s=100)
plt.title('Heart Rate (bpm) over Time Intervals')
plt.xlabel('Time Interval (min)')
plt.ylabel('Pulse (bpm)')

plt.subplot(2, 2, 3)
sns.histplot(data=df, x='pulse', color='teal', kde=True)
plt.title('Distribution of Heart Rates (bpm)')
plt.xlabel('Pulse (bpm)')
plt.ylabel('Count (Frequency)')

plt.subplot(2, 2, 4)
sns.boxplot(data=df, x='kind', y='pulse', palette='Set2')
plt.title('Heart Rate Range by Activity Intensity')
plt.xlabel('Activity Intensity')
plt.ylabel('Pulse (bpm)')

plt.tight_layout() 
plt.show()
```

```python
# Top Left (ID vs. Pulse): I chose these to see if individual participants had baseline heart rate differences. I learned that regardless of the person, the "Running" activity (purple/blue dots) almost always pushes the pulse higher than resting or walking.

# Top Right (Time vs. Pulse): I wanted to see if people's hearts slowed down or sped up over 30 minutes. I learned that for high-intensity exercise (Running), the heart rate stays elevated or climbs over the three time intervals, while resting stays flat.

# Bottom Left (Histogram of Pulse): I chose this to see the "average" heart rate of the group. I learned that most participants stay between 80 and 100 bpm, representing the majority of resting and walking data.

# Bottom Right (Boxplot of Activity): This was chosen to compare the spread of data. I learned that "Running" has a massive range (some people hit 150 bpm while others stay at 110 bpm), whereas "Resting" is very consistent for everyone.
```
