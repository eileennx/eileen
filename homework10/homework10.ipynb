```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
actual_path = r'C:\Users\eilee\python_decal_sp26\course_assignments\homework9\solar_system.csv'
df_raw = pd.read_csv(actual_path)

#a&b
print(f"Shape before transposing: {df_raw.shape}")
print(f"Rows before: {df_raw.shape[0]}, Columns before: {df_raw.shape[1]}")

#c
df = df_raw.set_index("Attribute").T
df.index.name = "Planet"
df.reset_index(inplace=True)
df.columns.name = None

#a&b
print(f"\nShape after transposing: {df.shape}")
print(f"Rows after: {df.shape[0]}, Columns after: {df.shape[1]}")

#c
print("\ndf.set_index('Attribute'): Moves the 'Attribute' column to be the row labels.")
print(".T: Flips the table so rows -> cols and cols -> rows.")
print("df.index.name = 'Planet': Labels the new row index as 'Planet'.")
print("df.reset_index(inplace=True): Converts the 'Planet' index back into a regular col")
print("df.columns.name = None: Removes the 'Attribute' label from the top of the cols")

#d
print("\nThe shape isn't (11, 20) because reset_index() adds 'Planet' as a new column, increasing the column count from 20 to 21.")

#e
print("\nNew column names:")
print(df.columns.tolist())

#f
units_cols = [c for c in df.columns if '(' in c]
no_units_cols = [c for c in df.columns if '(' not in c]
print(f"\nColumns with units ({len(units_cols)}): {units_cols}")
print(f"\nColumns without units ({len(no_units_cols)}): {no_units_cols}\n")

#g
print(df)
```

```python
# ---2.1.2 Investigate Data Types ---
#1
print(df.dtypes)

#2
for col in df.columns:
    print(f"{col}: {df[col].apply(type).unique()}")

#3
#3a
print("\nThe actual values are strings (<class 'str'>).")
#3b
print("\nStoring numeric values as strings is a problem because you cannot perform mathematical operations or create numerical plots.")
```

```python
# --- 2.1.3 Convert Strings to Numbers
#1
exclude = ["Planet", "Ring System?", "Global Magnetic Field?"]

for col in df.columns:
    if col not in exclude:
        df[col] = pd.to_numeric(df[col], errors="coerce")

#2
print(df.dtypes)

# 3
#3a
print("\nThe three different data types are object, float64, and int64.")
#3b
print("\n- float64: 64-bit floating point numbers used for decimals")
print("- int64: 64-bit integers used for whole numbers")
print("- object: A general pandas type used here for strings.")
#3c
print("\n'Planet', 'Ring System?', and 'Global Magnetic Field?' are still object because they contain text data that cannot be turned into numbers.")
```

```python
# --- 2.2 Apply Astropy Units ---
#1 
def attach_units(old_col, new_col, unit_multiplier):
    new_values = []
    for val in df[old_col]:
        new_values.append(val * unit_multiplier)
    
    df[old_col] = new_values
    df.rename(columns={old_col: new_col}, inplace=True)

#2, diameter
attach_units("Diameter (km)", "Diameter (km_unit)", 1 * u.km)
print(df["Diameter (km_unit)"])
```

```python
# --- 2.3 Data Analysis ---
attach_units("Perihelion (10^6 km)", "Perihelion (km)", 1e6 * u.km)
attach_units("Aphelion (10^6 km)", "Aphelion (km)", 1e6 * u.km)

semi_major_axis = (df["Perihelion (km)"] + df["Aphelion (km)"]) / 2

if "Semi-Major Axis (km)" not in df.columns:
    idx = df.columns.get_loc("Aphelion (km)") + 1
    df.insert(idx, "Semi-Major Axis (km)", semi_major_axis)

print("Updated Column Names:")
print(df.columns.tolist())
print("\nNew Semi-Major Axis Column:")
print(df["Semi-Major Axis (km)"])
```

```python
attach_units("Orbital Period (days)", "Period (days)", 1 * u.day)

df["Period (years)"] = [val.to(u.year) for val in df["Period (days)"]]
df.drop(columns=["Period (days)"], inplace=True)

print("Columns after Period conversion:")
print(df.columns.tolist())
print("\nFull Period (years) column:")
print(df["Period (years)"])

fav = "Mars"
fav_period = df.loc[df["Planet"] == fav, "Period (years)"].values[0]
print(f"\nMy favorite planet is {fav}, and its orbital period is {fav_period.value:.4f} years.")
```

```python
print("AU Definition: The average distance between the Earth and the Sun.\n")

print(f"One AU constant: {const.au}")

au_in_km = const.au.to(u.km)
print(f"\nThere are {au_in_km} in one AU.")

for col in list(df.columns):
    if "(km)" in col:
        new_name = col.replace("(km)", "(AU)")
        
        new_vals = []
        for val in df[col]:
            if hasattr(val, 'unit'):
                new_vals.append(val.to(u.AU))
            else:
                new_vals.append((val * u.km).to(u.AU))
        
        df[new_name] = new_vals
        df.drop(columns=[col], inplace=True)

print(f"\nSample AU values for {fav}:")
for col in df.columns:
    if "(AU)" in col:
        val = df.loc[df["Planet"] == fav, col].values[0]
        print(f"{col}: {val.value:.6f} AU")
```
