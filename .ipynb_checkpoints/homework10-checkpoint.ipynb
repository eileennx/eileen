```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
# --- 2.1.1 Fix Formatting ---
df = pd.read_csv('../../python_decal_sp26/course_assignments/homework9/solar_system.csv')

#5a
print(f"Shape before transposing: {df.shape}")
#5b
print(f"There are {df.shape[0]} rows and {df.shape[1]} columns before.\n")

#4
df = df.set_index("Attribute").T
df.index.name = "Planet"
df.reset_index(inplace=True)
df.columns.name = None

#5a
print(f"Shape after transposing: {df.shape}")

#5b
print(f"There are {df.shape[0]} rows and {df.shape[1]} columns after.\n")

#5c
print("""1. df.set_index("Attribute").T: Sets 'Attribute' as the row labels then swaps rows and columns.
2. df.index.name = "Planet": Names the new row index 'Planet'.
3. df.reset_index(inplace=True): Moves the 'Planet' index into a column and adds numbers for rows.
4. df.columns.name = None: Deletes the old 'Attribute' label from the top of the columns.
""")

#5d 
print("""The shape is (10, 21) because there are 10 planets (rows). We had 20 attributes that became columns, 
plus one extra column for the Planet names created by reset_index.
""")

#5e
print(df.columns.tolist())
print()

#5f
cols_units = [c for c in df.columns if '(' in c]
cols_no_units = [c for c in df.columns if '(' not in c]

print(f"5(f) {len(cols_units)} columns have units: {cols_units}")
print(f"{len(cols_no_units)} columns don't have units: {cols_no_units}")
print()

# 5g
print("5(g) Transposed DataFrame:")
df
```

```python
# --- 2.1.2 Investigate Data Types ---
#1
print("Current column dtypes:")
print(df.dtypes)

#2 
print("\nIndividual value types (first 3 columns example):")
for col in df.columns:
    print(f"{col}: {df[col].apply(type).unique()}")

#3
print("""
3(a) The actual values are currently strings (<class 'str'>).
3(b) Numeric values stored as strings cannot be used for mathematical operations 
     (like addition or taking an average) or for plotting numerical axes.
""")
```

```python
# --- 2.1.3 Convert Strings to Numbers ---
#1
cols_to_convert = [col for col in df.columns if col not in ["Planet", "Ring System?", "Global Magnetic Field?"]]

for col in cols_to_convert:
    df[col] = pd.to_numeric(df[col], errors="coerce")

#2
print("Data types after conversion:")
print(df.dtypes)

#3
print("""
3(a) The three data types are: object, float64, and int64.
3(b) Definitions:
     - float64: 64-bit floating point numbers (decimals).
     - int64: 64-bit integers (whole numbers).
     - object: A general pandas type usually representing strings (text) in this case.
3(c) 'Planet', 'Ring System?', and 'Global Magnetic Field?' are still object columns 
     because they contain text names (e.g., 'Earth') or categories ('Yes'/'No') 
     that do not have a numerical equivalent.""")
```

```python
#1
def attach_units(old_col, new_col, unit_multiplier):
    if old_col in df.columns:
        new_values = []
        for val in df[old_col]:
            new_values.append(val * unit_multiplier)
        
        df[old_col] = new_values
        df.rename(columns={old_col: new_col}, inplace=True)
        print(f"Success: '{old_col}' is now '{new_col}'")
    else:
        print(f"Skipped: '{old_col}' not found. (It might already be '{new_col}')")

#2
mass_unit = 1e24 * u.kg
attach_units("Mass (10^24kg)", "Mass (kg)", mass_unit)

print("\nColumn values after applying units:")
print(df["Mass (kg)"].head())
```

```python
# --- 2.3 Data Analysis ---
#1
attach_units("Perihelion (10^6 km)", "Perihelion (km)", 1e6 * u.km)
attach_units("Aphelion (10^6 km)", "Aphelion (km)", 1e6 * u.km)
semi_major_axis = (df["Perihelion (km)"] + df["Aphelion (km)"]) / 2
col_name = "Semi-Major Axis (km)"
if col_name not in df.columns:
    idx = df.columns.get_loc("Aphelion (km)") + 1
    df.insert(idx, col_name, semi_major_axis)
else:
    df[col_name] = semi_major_axis

print("Column names:", df.columns.tolist())
print("\nNew Semi-Major Axis values:")
print(df[col_name].head())
print()
```

```python
#2
attach_units("Orbital Period (days)", "Period (days)", 1 * u.day)

if "Period (days)" in df.columns:
    df["Period (years)"] = [val.to(u.year) for val in df["Period (days)"]]
    df.drop(columns=["Period (days)"], inplace=True)

print("Column names:", df.columns.tolist())
print("\nOrbital Period in Years:")
print(df["Period (years)"])

fav = "Earth"
fav_period = df.loc[df["Planet"] == fav, "Period (years)"].values[0]
print(f"\nMy favorite planet is {fav}, and its orbital period is {fav_period.value:.4f} years.")
print()
```

```python
#3
print("AU Definition: An Astronomical Unit (AU) is roughly the average distance from the Earth to the Sun.")

print(f"One AU with Astropy: {const.au}")

au_in_km = const.au.to(u.km)
print(f"There are {au_in_km} in one AU.")

for col in list(df.columns):
    if "(km)" in col:
        new_name = col.replace("(km)", "(AU)")
        
        new_values = []
        for val in df[col]:
            if hasattr(val, 'unit'):
                new_values.append(val.to(u.AU))
            else:
                new_values.append((val * u.km).to(u.AU))
        
        df[new_name] = new_values
        df.drop(columns=[col], inplace=True)

fav_planet = "Earth"
print(f"\nSample AU values for {fav_planet}:")
for col in df.columns:
    if "(AU)" in col:
        val = df.loc[df["Planet"] == fav_planet, col].values[0]
        print(f"{col}: {val.value:.6f} AU")
```

```python
# --- 2.4 Save Your Results ---
df.to_csv("units.csv", index=False)
```
